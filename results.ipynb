```python
import json
from pathlib import Path
import pandas as pd

energy_summary_paths = [
    "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260729_qwen36_27b_n100/energy_summary.json",
    "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260725_mistral_small_24b_n100/energy_summary.json",
    "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260727_gemma3_27b_n100/energy_summary.json",
]

def get_nested(d, *keys):
    for k in keys:
        if not isinstance(d, dict) or k not in d:
            return None
        d = d[k]
    return d

rows = []
for path in energy_summary_paths:
    path = Path(path)
    with open(path, "r") as f:
        data = json.load(f)

    run_name = path.parent.name

    eval_path = path.parent / "eval_report.json"
    eval_data = json.load(open(eval_path)) if eval_path.exists() else {}

    total_gpu_energy_wh = get_nested(data, "total", "total_gpu_energy_wh")
    net_energy_wh = get_nested(data, "total", "net_energy_wh")
    resolved_instances = eval_data.get("resolved_instances")
    completed_instances = eval_data.get("completed_instances")
    n_limit = eval_data.get("n_limit")

    rows.append({
        "run": run_name,
        "idle_baseline_watts": get_nested(data, "idle_baseline_watts"),

        "total_duration_s": get_nested(data, "total", "duration_seconds"),
        "total_avg_gpu_power_w": get_nested(data, "total", "avg_gpu_power_watts"),
        "total_max_gpu_power_w": get_nested(data, "total", "max_gpu_power_watts"),
        "total_gpu_energy_wh": total_gpu_energy_wh,
        "net_energy_wh": net_energy_wh,
        "total_samples": get_nested(data, "total", "samples"),

        "setup_duration_s": get_nested(data, "phases", "setup", "duration_seconds"),
        "setup_avg_gpu_power_w": get_nested(data, "phases", "setup", "avg_gpu_power_watts"),
        "setup_energy_wh": get_nested(data, "phases", "setup", "energy_wh"),

        "inference_duration_s": get_nested(data, "phases", "inference", "duration_seconds"),
        "inference_avg_gpu_power_w": get_nested(data, "phases", "inference", "avg_gpu_power_watts"),
        "inference_energy_wh": get_nested(data, "phases", "inference", "energy_wh"),
        "inference_net_energy_wh": get_nested(data, "phases", "inference", "net_energy_wh"),

        "per_instance_count": get_nested(data, "per_instance_count"),

        # recalculées à partir de eval_report.json (couvre aussi qwen, absent du fichier source)
        "energy_per_resolved_wh": round(total_gpu_energy_wh / resolved_instances, 1) if resolved_instances else None,
        "net_energy_per_resolved_wh": round(net_energy_wh / resolved_instances, 1) if resolved_instances else None,
        "energy_per_evaluated_instance_wh": round(total_gpu_energy_wh / completed_instances, 1) if completed_instances else None,
        "energy_per_attempted_instance_wh": round(total_gpu_energy_wh / n_limit, 1) if n_limit else None,
    })

df = pd.DataFrame(rows).sort_values("run").reset_index(drop=True)
pd.set_option("display.max_columns", None)
df
```

```python
import json
from pathlib import Path
import pandas as pd

tool_summary_paths = {
    "qwen36_27b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260729_qwen36_27b_n100/tool_summary.json",
    "gemma3_27b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260727_gemma3_27b_n100/tool_summary.json",
    "mistral_small_24b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260725_mistral_small_24b_n100/tool_summary.json",
}

data_by_model = {model: json.load(open(path)) for model, path in tool_summary_paths.items()}

# top-level metrics
top_rows = []
for model, data in data_by_model.items():
    top_rows.append({
        "model": model,
        "total_tool_calls": data.get("total_tool_calls"),
        "total_instances": data.get("total_instances"),
        "avg_tool_calls_per_instance": data.get("avg_tool_calls_per_instance"),
    })
df_top = pd.DataFrame(top_rows).sort_values("model").reset_index(drop=True)

# tool-level breakdown (counts + pct), union of all tool names seen across models
all_tools = sorted(set().union(*(data.get("global_tool_counts", {}).keys() for data in data_by_model.values())))

tool_rows = []
for tool in all_tools:
    row = {"tool": tool}
    for model, data in data_by_model.items():
        row[f"count_{model}"] = data.get("global_tool_counts", {}).get(tool)
        row[f"pct_{model}"] = data.get("tool_distribution_pct", {}).get(tool)
    tool_rows.append(row)
df_tools = pd.DataFrame(tool_rows)

pd.set_option("display.max_columns", None)
df_top
```

```python
import json
from pathlib import Path
import pandas as pd



per_instance_paths = {
    "qwen36_27b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260729_qwen36_27b_n100/per_instance_full.json",
    "gemma3_27b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260727_gemma3_27b_n100/per_instance_full.json",
    "mistral_small_24b": "/Users/z338mn/Library/CloudStorage/OneDrive-AXA/Cursor/OpenHands_Benchmark/results/20260725_mistral_small_24b_n100/per_instance_full.json",
}

resolved_ids = {
    "qwen36_27b": {
        "astropy__astropy-12907","astropy__astropy-14995","django__django-10914",
        "django__django-10924","django__django-11001","django__django-11039",
        "django__django-11049","django__django-11099","django__django-11133",
        "django__django-11179","django__django-11583","django__django-11620",
        "django__django-11999","django__django-12125","django__django-12184",
        "django__django-12284","django__django-12286","django__django-12453",
        "django__django-12497","django__django-12700","django__django-12908",
        "django__django-12915","django__django-12983","django__django-13028",
        "django__django-13158","django__django-13230","django__django-13315",
        "django__django-13447","django__django-13590","django__django-13658",
        "django__django-13710","django__django-13757","django__django-13933",
        "django__django-14016","django__django-14238","django__django-14382",
        "django__django-14411","django__django-14580","django__django-14608",
        "django__django-14672","django__django-14752","django__django-14787",
        "django__django-14855","django__django-14915","django__django-14999",
        "django__django-15347","django__django-15498","django__django-15695",
        "django__django-15789","django__django-15790","django__django-15814",
    },
    "gemma3_27b": {
        "django__django-11039","django__django-11099","django__django-13658",
        "django__django-14382","django__django-14672","django__django-15061",
        "django__django-15347","django__django-15814",
    },
    "mistral_small_24b": {
        "django__django-11039","django__django-11099","django__django-11133",
        "django__django-12286","django__django-14382","django__django-14672",
        "django__django-14855","django__django-14915","django__django-15347",
        "django__django-15789",
    },
}

METRICS = ["completion_tokens", "prompt_tokens", "energy_wh", "avg_power_watts", "duration_seconds"]

tokens_by_model = {}
for model, path in per_instance_paths.items():
    data = json.load(open(path))
    tokens_by_model[model] = {
        row["instance_id"]: {m: row.get(m) for m in METRICS}
        for row in data["per_instance"]
    }

models = list(per_instance_paths.keys())

def build_comparison(instance_ids, models_subset=models):
    rows = []
    for instance_id in sorted(instance_ids):
        solvers = [m for m in models_subset if instance_id in resolved_ids[m]]
        row = {"instance_id": instance_id, "solved_by": ",".join(solvers), "n_solved": len(solvers)}
        for m in METRICS:
            for model in models_subset:
                vals = tokens_by_model[model].get(instance_id, {})
                row[f"{m}_{model}"] = vals.get(m)
        rows.append(row)
    df = pd.DataFrame(rows)
    pd.set_option("display.max_columns", None)
    return df

# instances résolues par les 3 modèles
common_all = set.intersection(*resolved_ids.values())
df_common_all = build_comparison(common_all)

# instances résolues par au moins 2 modèles
solve_count = {}
for model, ids in resolved_ids.items():
    for instance_id in ids:
        solve_count[instance_id] = solve_count.get(instance_id, 0) + 1
common_at_least_2 = {i for i, c in solve_count.items() if c >= 2}
df_common_at_least_2 = build_comparison(common_at_least_2)
df_common_at_least_2 = df_common_at_least_2.sort_values(
    ["n_solved", "instance_id"], ascending=[False, True]
).reset_index(drop=True)

df_common_at_least_2
```

```python
resolved_ids_by_run = {}
for path in eval_report_paths:
    path = Path(path)
    with open(path, "r") as f:
        data = json.load(f)
    resolved_ids_by_run[path.parent.name] = set(data.get("resolved_ids", []))

all_runs = list(resolved_ids_by_run.keys())

solved_by_any = set.union(*resolved_ids_by_run.values())

solve_count = {
    instance_id: sum(instance_id in ids for ids in resolved_ids_by_run.values())
    for instance_id in solved_by_any
}

matrix_rows = []
for instance_id in sorted(solved_by_any):
    row = {"instance_id": instance_id}
    for run in all_runs:
        row[run] = "✓" if instance_id in resolved_ids_by_run[run] else ""
    row["n_solved"] = solve_count[instance_id]
    matrix_rows.append(row)

df_matrix = pd.DataFrame(matrix_rows).sort_values("n_solved", ascending=False).reset_index(drop=True)
pd.set_option("display.max_columns", None)
df_matrix
```
